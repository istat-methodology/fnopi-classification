```python
import json
import os
from dotenv import load_dotenv

from openai import OpenAI
```

```python
OPENAI_MODEL = "gpt-5-mini" #gpt-5-mini, gpt-4o-mini 
SYSTEM_PROMPT = """Sei un esperto di minfermieristica ."""
USER_INSTRUCTIONS = """Assegna una categoria al  prompt che è un diario infermieristico basata su queste categorie: 
assessment
valutazione dolore
monitoraggio drenaggio
posizionato/riposizionato cvp
esami ematici
medicazione cvc
assessment, valutazione dolore
emocolture
rilevata glicemia
gestione cvc
tampone antigenico
tampone covid
catetere arterioso radiale (gest/med)
rimosso cvc
tampone rapido
emogasanalisi (ega)
gestione elastomero
medicazione compressiva
medicazione lesione sovraombelicale
posizionamento monitor
posizionato mad
presenza della figlia in camera, senza permesso e senza requisiti di sicurezza
rimosso cvc radiale
rimosso cvp
sostituito cvp
tampone
tampone faringeo
urinocoltura
valutazione del dolore: monitoraggio cvc.

 Puoi dare 0, 1 o 2 classificazioni per il diario nel prompt.
 L'output deve essere:
' 1- Categoria, 2-categoria'
oppure :
'no classification'  """


```

```python
# Chiave API
load_dotenv()
api_key = os.getenv("OPENAI_API_KEY_CHAT")
client = OpenAI(api_key=api_key)
```

```python
testo_diario="""RILEVATI PV NRS 0 TA 37.5TAMPONE COVID NEGATIVO SOMMINISTRATA TERAPIA E. TA ASSISTENZA DOMANI MATTINA ESEGUE TACPZ SI RIMUOVE SNG----> 400 CC"""


```

```python
resp = client.responses.create(
	model=OPENAI_MODEL,
	instructions=SYSTEM_PROMPT,
	input=USER_INSTRUCTIONS + testo_diario + "\n---"
)
print(resp.output_text) 
# data = json.loads(resp.output_text)
# print(json.dumps(data, indent=4, ensure_ascii=False))
```

```python
for i,text in enumerate(lista_diari[:15]):
    resp = client.responses.create(
	model=OPENAI_MODEL,
	instructions=SYSTEM_PROMPT,
	input=USER_INSTRUCTIONS + text + "\n---"
    )
    
    print('output:\t ',resp.output_text)
    print('verità:\t ',df['intervento_sicurezza'].iloc[i], '\n')

# data = json.loads(resp.output_text)
# print(json.dumps(data, indent=4, ensure_ascii=False))
```

```python
df_expl_copy = df_exploded.copy()
```

```python
# colonna di partenza (già normalizzata a minuscolo/spazi puliti)
col = "intervento_sicurezza_list"

mapping_macro = {
    # valutazione / monitoraggio clinico
    "assessment": "valutazione/monitoraggio clinico",
    #"valutazione dolore": "valutazione/monitoraggio clinico",
    "assessment, valutazione dolore": "valutazione/monitoraggio clinico",
    "valutazione del dolore: monitoraggio cvc": "valutazione/monitoraggio clinico",

    # drenaggi
    "monitoraggio drenaggio": "drenaggi",

    # diagnostica lab/POC (collassata)
    "esami ematici": "diagnostica lab/POC",
    "rilevata glicemia": "diagnostica lab/POC",
    "emogasanalisi (ega)": "diagnostica lab/POC",
    "urinocoltura": "diagnostica lab/POC",

    # emocolture (tenuta a parte)
    "emocolture": "emocolture",

    # tamponi (solo covid / altro)
    "tampone covid": "tamponi – covid",
    "tampone antigenico": "tamponi – covid",
    #"tampone rapido": "tamponi – covid",
    "tampone": "tamponi – altro",
    "tampone faringeo": "tamponi – altro",

    # accessi venosi (cvc/cvp)
    "medicazione cvc": "accessi venosi (cvc/cvp)",
    "gestione cvc": "accessi venosi (cvc/cvp)",
    "rimosso cvc": "accessi venosi (cvc/cvp)",
    "posizionato/riposizionato cvp": "accessi venosi (cvc/cvp)",
    "rimosso cvp": "accessi venosi (cvc/cvp)",
    "sostituito cvp": "accessi venosi (cvc/cvp)",

    # accessi arteriosi
    "catetere arterioso radiale (gest/med)": "accessi arteriosi",
    "rimosso cvc radiale": "accessi arteriosi",  # se nel tuo contesto è effettivamente arterioso

    # cura ferite / medicazioni
    "medicazione compressiva": "cura ferite/medicazioni",
    "medicazione lesione sovraombelicale": "cura ferite/medicazioni",

    # device / terapie varie
    "gestione elastomero": "device/terapie varie",
    "posizionato mad": "device/terapie varie",

    # strumentazione / monitor
    "posizionamento monitor": "strumentazione/monitor",

    # eventi non clinici / sicurezza
    "presenza della figlia in camera, senza permesso e senza requisiti di sicurezza": "eventi non clinici/sicurezza",
}

# applica: niente "non classificato"; se manca nel mapping, lascia la voce così com'è
df_expl_copy["macro_cat"] = (
    df_expl_copy[col]
    .str.strip().str.lower().replace(r"\s+", " ", regex=True)
    .map(mapping_macro)
    .fillna(df_expl_copy[col])   # mantiene l'etichetta originale se non mappata
)

# esempio conteggi
# df_exploded["macro_cat"].value_counts()
```
