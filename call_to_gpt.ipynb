# GPT

```python
import json
import os
from dotenv import load_dotenv

from openai import OpenAI
```

```python
OPENAI_MODEL = "gpt-5-mini" #gpt-5-mini, gpt-4o-mini 
SYSTEM_PROMPT = """Sei un esperto di minfermieristica ."""
USER_INSTRUCTIONS = """Assegna una categoria al  prompt che è un diario infermieristico basata su queste categorie: 
assessment
valutazione dolore
monitoraggio drenaggio
posizionato/riposizionato cvp
esami ematici
medicazione cvc
assessment, valutazione dolore
emocolture
rilevata glicemia
gestione cvc
tampone antigenico
tampone covid
catetere arterioso radiale (gest/med)
rimosso cvc
tampone rapido
emogasanalisi (ega)
gestione elastomero
medicazione compressiva
medicazione lesione sovraombelicale
posizionamento monitor
posizionato mad
presenza della figlia in camera, senza permesso e senza requisiti di sicurezza
rimosso cvc radiale
rimosso cvp
sostituito cvp
tampone
tampone faringeo
urinocoltura
valutazione del dolore: monitoraggio cvc.

 Puoi dare 0, 1 o 2 classificazioni per il diario nel prompt.
 L'output deve essere:
' 1- Categoria, 2-categoria'
oppure :
'no classification'  """


```

```python
# Chiave API
load_dotenv()
api_key = os.getenv("OPENAI_API_KEY_CHAT")
client = OpenAI(api_key=api_key)
```

```python
testo_diario="""RILEVATI PV NRS 0 TA 37.5TAMPONE COVID NEGATIVO SOMMINISTRATA TERAPIA E. TA ASSISTENZA DOMANI MATTINA ESEGUE TACPZ SI RIMUOVE SNG----> 400 CC"""


```

```python
resp = client.responses.create(
	model=OPENAI_MODEL,
	instructions=SYSTEM_PROMPT,
	input=USER_INSTRUCTIONS + testo_diario + "\n---"
)
print(resp.output_text) 
# data = json.loads(resp.output_text)
# print(json.dumps(data, indent=4, ensure_ascii=False))
```

```python
for i,text in enumerate(lista_diari[:15]):
    resp = client.responses.create(
	model=OPENAI_MODEL,
	instructions=SYSTEM_PROMPT,
	input=USER_INSTRUCTIONS + text + "\n---"
    )
    
    print('output:\t ',resp.output_text)
    print('verità:\t ',df['intervento_sicurezza'].iloc[i], '\n')

# data = json.loads(resp.output_text)
# print(json.dumps(data, indent=4, ensure_ascii=False))
```

```python

df["intervento_sicurezza_list"] = df["intervento_sicurezza"].str.split(";")
df_exploded = df.explode("intervento_sicurezza_list")
df_exploded.head()
```

```python
df_exploded["intervento_sicurezza_list"].value_counts().to_csv("intervento_sicurezza_counts_1.csv")
```

```python
cols = df_exploded.columns.tolist()

col = "intervento_sicurezza_list"

# togli la colonna dalla lista e reinseriscila in posizione 2 (che è la terza, contando da 0)
cols.remove(col)
cols.insert(7, col)

# riordina il DataFrame
df_exploded = df_exploded[cols]
df_exploded.head(5)
```

```python
df_exploded["intervento_sicurezza_list"].value_counts()

```

```python
# pulire righe con spazi e maiuscole minuscole :"intervento_sicurezza_list"

df_exploded["intervento_sicurezza_list"] = (
    df_exploded["intervento_sicurezza_list"]
    .str.strip()        # toglie spazi in eccesso
    .str.lower()        # tutto minuscolo
)

```

```python
df_exploded["intervento_sicurezza_list"].value_counts().to_csv("intervento_sicurezza_counts_2.csv")
```

```python
# mappings di cose uguali
col = "intervento_sicurezza_list"

mapping_conserv = {
    "asssessment": "assessment",
    "valutazione del dolore": "valutazione dolore",
    "monitoraggio drenaggi": "monitoraggio drenaggio",
    "rimosso il cvc": "rimosso cvc",
    "assesment, valutazione dolore" : "assessment, valutazione dolore"
}
df_exploded[col] = df_exploded[col].replace(mapping_conserv)


```

```python
df_exploded["intervento_sicurezza_list"].value_counts()
```

```python
df_exploded["intervento_sicurezza_list"].value_counts().to_csv("intervento_sicurezza_counts_3.csv")
```

```python
# etichette piu comprensive e pulite, mapping finale:

mapping = {
    # dolore / assessment
    "valutazione del dolore": "valutazione dolore",
    "valutaizone del dolore": "valutazione dolore",
    "assesment": "assessment",
    "asssessment": "assessment",
    # drenaggi
    "monitoraggio drenaggi": "monitoraggio drenaggio",
    # esami ematici
    "eseguiti esami ematici": "esami ematici",
    "esegue esami ematici": "esami ematici",
    # glicemia
    "hgt": "rilevata glicemia",
    # cvp/cvc
    "riposizionato cvp": "posizionato/riposizionato cvp",
    "posizionato cvp": "posizionato/riposizionato cvp",
    "medicazione compressiva cvc": "medicazione cvc",
    "medicazione emostatica cvc": "medicazione cvc",
    "medicazione sede cvc": "medicazione cvc",
    "cvc": "gestione cvc",
    # catetere arterioso radiale
    "gestione catetere arterioso radiale": "catetere arterioso radiale (gest/med)",
    "medicazione catetere arterioso radiale": "catetere arterioso radiale (gest/med)",
    # emocolture
    "emocolture da cvc e cvp": "emocolture",
    "emocolture da picc e cvp": "emocolture",
    "eseguite emocolture": "emocolture",
    # tamponi (macro)
    "test antigenico per covid": "tampone covid",
    "eseguito tampone": "tampone",
    "eseguito tampone rapido": "tampone rapido",

    # altro
    "ega": "emogasanalisi (EGA)",
}

# applica
col = "intervento_sicurezza_list"
df_exploded[col] = df_exploded[col].str.strip().str.lower().replace(r"\s+", " ", regex=True)
df_exploded[col] = df_exploded[col].replace(mapping)

# NB: per le voci combinate va prima fatto lo split:
# "assessment, valutazione dolore" -> split su virgola o ":" prima di replace


```
