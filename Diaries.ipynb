```python
import pandas as pd
from bs4 import BeautifulSoup
from html import unescape
```

```python
df_diari = pd.read_csv("diaries/Esempio_diari.csv", sep=';',encoding='latin-1')
```

```python
df_diari.columns
```

```python
df_diari['testo diario'].head()
```

```python
def drop_html(html_text):

    # Parsing HTML
    soup = BeautifulSoup(html_text, 'html.parser')

    # Rimozione dei tag HTML
    clean_text = soup.get_text(separator=" ", strip=True)

    # Conversione entità HTML (&nbsp;, &egrave;, ecc.)

    clean_text = unescape(clean_text)

    return clean_text
```

```python
df_diari['testo_diario_clean'] =  df_diari["testo diario"].apply(lambda x: drop_html(x))
```
